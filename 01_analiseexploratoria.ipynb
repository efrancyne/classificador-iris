```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

## Importando dados


```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
df["Species"] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

## Relação entre Comprimento e Largura das Pétalas

```python
sns.scatterplot(
    data=df,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="Species",
    palette="Set2"
)

plt.title("Distribuição das espécies pelas características das pétalas")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="sepal width (cm)",
    hue="Species",
    palette="Set2"
)

plt.title("Distribuição das espécies pelas características das sépalas")
plt.show()
```

## Distribuição do Comprimento das Pétalas por Espécie

```python
sns.boxplot(df, x='Species',y='petal length (cm)')
plt.grid()
plt.show()
```
